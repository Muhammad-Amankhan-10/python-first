```python
num = int(input("Enter first number: "))

with open("multiplication_table.txt", "w+") as file:
    for i in range(1, 11):
        file.write(f"{num} x {i} = {num * i}\n")

with open("multiplication_table.txt") as file:
    print(file.read())
   
```
